```python
bins = [0, 5000, 15000, df['price'].max()+1]
labels = ['Low', 'Medium', 'High']
df['price_category'] = pd.cut(df['price'], bins=bins, labels=labels, right=False)

bins_host = [1, 15, 90, df['Number_of_properties'].max()+1]
labels_host = ['Basic', 'Experienced', 'Professional']
df['host_experience'] = pd.cut(df['Number_of_properties'], bins=bins_host, labels=labels_host, right=False)

bins_reviews = [0, 30, 90, df['number_of_reviews'].max()+1]
labels_reviews = ['Basic', 'Recommended', 'Popular']
df['review_category'] = pd.cut(df['number_of_reviews'], bins=bins_reviews, labels=labels_reviews, right=False)

bins_availability = [0, 20, 150, df['availability_365'].max()+1]
labels_availability = ['Not available', 'available', 'Highly available']
df['availability_category'] = pd.cut(df['availability_365'], bins=bins_availability, labels=labels_availability, right=False)

print(df.head())
```

5 כיצד משתנים אורכי השהייה של השכרות Airbnb לפי שכונה? האם יש שכונות שנוטות
למשוך שהיות קצרות או ארוכות יותר?

```python
import pandas as pd
import matplotlib.pyplot as plt

# לא לדרוס את df המקורי + המרה למספרי
tmp = df.copy()
tmp['minimum_nights'] = pd.to_numeric(tmp['minimum_nights'], errors='coerce')

# סינון חריגים מאוד (עד שנה), והתעלמות מ-NaN
tmp = tmp[(tmp['minimum_nights'] < 365) & (tmp['minimum_nights'].notna())]

# לקחת את 12 השכונות עם הכי הרבה נכסים כדי שיהיה קריא
top_neigh = tmp['neighbourhood'].value_counts().head(12).index
stay_by_neighborhood = (tmp[tmp['neighbourhood'].isin(top_neigh)]
                        .groupby('neighbourhood')['minimum_nights']
                        .mean().sort_values())

# גרף
plt.figure(figsize=(12,6))
plt.bar(stay_by_neighborhood.index, stay_by_neighborhood.values, color='skyblue')
plt.title('Average minimum number of nights by neighborhood (12 Popularity)')
plt.xlabel('neighborhood'); plt.ylabel('Minimum nights (average)')
plt.xticks(rotation=45, ha='right')
plt.grid(True, axis='y')
plt.tight_layout()
plt.show()
```

7 מצאו את מספר הביקורות הכולל ואת מספר הביקורות המרבי לפי כל קבוצת שכונות.

```python
reviews_by_neighbourhood = df.groupby('neighbourhood')['number_of_reviews'].sum().sort_values(ascending=False)

# חישוב גבולות שלישונים
n = len(reviews_by_neighbourhood)
top_third = reviews_by_neighbourhood.iloc[:n//3]
middle_third = reviews_by_neighbourhood.iloc[n//3: 2*n//3]
bottom_third = reviews_by_neighbourhood.iloc[2*n//3:]

# פונקציה לחישוב סך וביקורת מקסימלית
def calculate_stats(neighbourhoods):
    subset = df[df['neighbourhood'].isin(neighbourhoods.index)]
    total_reviews = subset['number_of_reviews'].sum()
    max_reviews = subset['number_of_reviews'].max()
    return total_reviews, max_reviews

# חישוב לכל שליש
top_total, top_max = calculate_stats(top_third)
middle_total, middle_max = calculate_stats(middle_third)
bottom_total, bottom_max = calculate_stats(bottom_third)

print("Top third of neighborhoods by number of reviews")
print(f"Total reviews in the top third: {top_total}")
print(f"Highest number of reviews in the top third: {top_max}\n")

print("Middle third of the neighborhoods")
print(f"Total reviews in the middle third: {middle_total}")
print(f"Highest number of reviews in the middle third: {middle_max}\n")

print("Bottom third of the neighborhoods")
print(f"Total reviews in the bottom third: {bottom_total}")
print(f"The highest number of reviews in the bottom third: {bottom_max}")

```

.8 מצאו את סוגי הנכס עם הכי הרבה ביקורות בקבוצות השכונות.

```python
def top_room_type(neighbourhoods):
    subset = df[df['neighbourhood'].isin(neighbourhoods.index)]
    room_counts = subset.groupby('room_type')['number_of_reviews'].sum()
    top_type = room_counts.idxmax()
    top_reviews = room_counts.max()
    return top_type, top_reviews

# חישוב לכל שליש
top_type, top_reviews = top_room_type(top_third)
middle_type, middle_reviews = top_room_type(middle_third)
bottom_type, bottom_reviews = top_room_type(bottom_third)

# הדפסת הנתונים
print("Top third:")
print(f"  Most reviewed room type: {top_type}")
print(f"  Number of reviews: {top_reviews}\n")

print("Middle third:")
print(f"  Most reviewed room type: {middle_type}")
print(f"  Number of reviews: {middle_reviews}\n")

print("Bottom third:")
print(f"  Most reviewed room type: {bottom_type}")
print(f"  Number of reviews: {bottom_reviews}\n")
```

9 מצאו את מיקום הנכס הטוב ביותר עבור מטיילים על פי קריטריונים שונים כגון מחיר זול,
שכונה מבוקשת, הכי הרבה ביקורות ועוד.

```python
# שלב 1: סך הביקורות לפי שכונה
reviews_by_neighbourhood = df.groupby('neighbourhood')['number_of_reviews'].sum().sort_values(ascending=False)

# בוחרים את השכונה הכי פופולרית
top_2_neighbourhoods = reviews_by_neighbourhood.head(1).index.tolist()

# שלב 2: חישוב ממוצעים
stats = df[df['neighbourhood'].isin(top_2_neighbourhoods)].groupby('neighbourhood').agg({
    'price': 'mean',
    'availability_365': 'mean',
    'number_of_reviews': 'sum'
}).sort_values(by='price')  # כדי לראות את המחיר הזול קודם

# שלב 3: הצגה ברורה
for idx, row in stats.iterrows():
    print(f"neighborhood: {idx}")
    print(f"Average price per property: {row['price']:.2f}")
    print(f"Total reviews: {row['number_of_reviews']}")
    print(f"Average availability per year (days): {row['availability_365']:.1f}\n")
```

0מצאו ממוצעת מחירים בין קבוצות השכונות, כמות נכסים ועוד.

```python
import matplotlib.pyplot as plt
import numpy as np

# חישוב סטטיסטיקות לפי שכונה
neighbourhood_stats = df.groupby('neighbourhood').agg({
    'price': 'mean',
    'id': 'count',  # מספר נכסים
    'availability_365': 'mean'
}).sort_values(by='price')  # אפשר למיין לפי מחיר ממוצע

# בחירת מספר השכונות שרוצים להציג, למשל 5 הכי פופולריות לפי מספר נכסים
top_neighbourhoods = neighbourhood_stats.sort_values(by='id', ascending=False).head(5)

# נתונים לגרף
labels = top_neighbourhoods.index.tolist()
price = top_neighbourhoods['price'].tolist()
num_properties = top_neighbourhoods['id'].tolist()
availability = top_neighbourhoods['availability_365'].tolist()

x = np.arange(len(labels))  # מיקום השכונות על ציר X
width = 0.25  # רוחב העמודה

# יצירת גרף
plt.figure(figsize=(12,6))
plt.bar(x - width, price, width, label='average price', color='skyblue')
plt.bar(x, num_properties, width, label='Amount of Assets', color='lightgreen')
plt.bar(x + width, availability, width, label='Average availability', color='salmon')

plt.xticks(x, labels, rotation=15)
plt.xlabel('average')
plt.ylabel('values')
plt.title('Comparing neighborhoods: price, number of properties and availability')
plt.legend()
plt.show()
```

11. מצאו אחוז נכסים לא פעילים פר שכונה– נכס לא פעיל הוא נכס שאין לו ביקורת בשנתיים
אחרונות ( ניתן לקחת שנה בשלימותה)

```python
import pandas as pd
import matplotlib.pyplot as plt

area = 'neighbourhood' if 'neighbourhood' in df.columns else 'City_Area'

# נוודא שהעמודה קיימת ומספרית
df['number_of_reviews_ltm'] = pd.to_numeric(df['number_of_reviews_ltm'], errors='coerce')

# נכס לא פעיל בשנתיים = לא היו ביקורות גם ב־12 חודשים אחרונים וגם ב־24 חודשים בערך (reviews_per_month==0)
if 'reviews_per_month' in df.columns:
    df['reviews_per_month'] = pd.to_numeric(df['reviews_per_month'], errors='coerce')
    inactive_mask = (df['number_of_reviews_ltm'].fillna(0) == 0) & (df['reviews_per_month'].fillna(0) == 0)
else:
    # אם אין לנו reviews_per_month → נשתמש רק ב-number_of_reviews_ltm כמדד משוער
    inactive_mask = (df['number_of_reviews_ltm'].fillna(0) == 0)

# חישוב אחוז לא פעילים לכל שכונה
inactive_pct = (inactive_mask.groupby(df[area]).mean() * 100).sort_values(ascending=False)

# גרף ל-12 שכונות מובילות
top = inactive_pct.head(12)
plt.figure(figsize=(12,6))
plt.bar(top.index.astype(str), top.values, color='salmon')
plt.title('Percentage of inactive properties in the last two years by neighborhood')
plt.xlabel('neighborhood')
plt.ylabel('Percentage of inactive people (%)')
plt.xticks(rotation=45, ha='right')
plt.grid(True, axis='y')
plt.tight_layout()
plt.show()
```
